```python
import random

# Prepare some data
names = ["Alice", "Bob", "Charlie", "Dave", "Eve"]
greetings = ["Hey", "Hi", "Hello", "Yo", "Howdy"]
questions = ["What's up?", "How are you?", "What are you doing?", "How's it going?"]
locations = ["Paris", "Tokyo", "Hawaii", "Rome", "New York City"]
activities = ["hiking", "sightseeing", "eating delicious food", "relaxing on the beach", "exploring the city"]

# Function to generate a random SMS message
def generate_message():
  # Randomly decide if this message is travel-related (adjust probability as needed)
  is_travel = random.random() < 0.05  

  if is_travel:
    name = random.choice(names)
    location = random.choice(locations)
    activity = random.choice(activities)
    message = f"{name}: Just arrived in {location}! {activity} is amazing!"
  else:
    greeting = random.choice(greetings)
    name = random.choice(names)
    question = random.choice(questions)
    message = f"{greeting} {name}, {question}"

  return message

# Generate 1000 messages
sms_messages = []
for _ in range(1000):
  sms_messages.append(generate_message())

# Print or save to a file
for message in sms_messages:
  print(message)

# To save to a file:
with open("sms_messages.txt", "w") as f:
    for message in sms_messages:
         f.write(message + "\n")
```

```python
from openai import OpenAI
import pandas as pd
import os
from dotenv import load_dotenv

load_dotenv()


client = OpenAI(api_key = os.getenv("OPENAI_API_KEY"))


def makeSMS( prompt):
    completion = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {
            "role": "user",
            "content": prompt
        }
    ]
    )
    #print(type(completion.choices[0].messages))
    return(completion.choices[0].message.content)
travelPrompt = "Generate a realistic SMS message that includes travel-related information please include sender and reciever."
financialPrompt = "Generate a realistic SMS message that includes a financial transaction please include sender and reciever."
otherPrompt = "Generate a realistic SMS message for general notifications or offers please include sender and reciever."

totalRecords = 15
travelCount = 0
finanicalCount = 0
otherCount = 0
promptFlag = 0
message = {}
messageTotal = pd.DataFrame()


for record in range (totalRecords):
    if(record % 10 == 0):
        msg = travelPrompt
        travelCount += 1
        promptFlag = 1
    elif(record % 12 == 0):
        msg =financialPrompt
        finanicalCount += 1
        promptFlag = 2
    else:
        msg=otherPrompt
        promptFlag = 0
        otherCount += 1
        
    msgReply = makeSMS(msg)
    message ={"Message":msgReply, "Type":promptFlag}   
    df_dictionary = pd.DataFrame([message])
    messageTotal = pd.concat([messageTotal, df_dictionary], ignore_index=True)

    print("Others: ", otherCount, " Financial:", finanicalCount, " Travel:", travelCount)
       
print(len(messageTotal))
messageTotal.to_csv
```
