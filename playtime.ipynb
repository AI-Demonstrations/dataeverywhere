```python
import re
def find_money(text):
    # Regular expression to match cash amounts
    pattern = r'\$\d{1,3}(?:,\d{3})*(?:\.\d{2})?'
    matches = re.findall(pattern, text)
    return matches
matches = find_money(" hello this is fred I just got $100 from susan and $1200 from  bill")
```
