## First determine type of message

```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification")
answer = classifier(
    "get $200 at the bank today",
    candidate_labels=["travel", "finance", "other"],
)
print(type(answer))
print(answer)
```

```python
print("Text:", answer["sequence"], " Category:", answer["labels"][0], " Certainity:", answer["scores"][0])
```

```python
results = {"Text": answer["sequence"], "Category": answer["labels"][0], "Certainity": answer["scores"][0]}
```

```python
print(results)
```
