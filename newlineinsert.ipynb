```python
# Function to insert newline after every period in a text file
def insert_newline_after_period(input_file, output_file):
    with open(input_file, 'r') as f:
        text = f.read()

    # Insert a newline after each period followed by a space or end of line
    modified_text = text.replace('. ', '.\n').replace('.', '.\n')

    with open(output_file, 'w') as f:
        f.write(modified_text)

# Specify the input and output file paths
input_file = '../data/data2'  # Replace with the path to your input file
output_file = '../data/data2output.txt'  # Replace with the path for the output file

# Call the function
insert_newline_after_period(input_file, output_file)
print(f"New file with inserted newlines saved to '{output_file}'")
```
